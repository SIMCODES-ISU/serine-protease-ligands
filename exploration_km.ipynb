```python
import numpy as np
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#log10km_mean vs value: See how input activator values affect catalytic turnover
import plotly.express as px

# Reload the CSV file for filtered_km_serine_proteases
km_df = pd.read_csv("/Users/chideraangwadom/SerineProtease/molecule_analysis/filtered_km_serine_proteases.csv")

fig1 = px.scatter(
    km_df,
    x="log10km_mean",
    y="value",
    title="log10km_mean vs Value"
    hover_data = ["substrate_smiles"]
)
fig1.show()
```

```python
#log10km_mean vs log10_value: Stabilized version of above; better for linear modeling
fig2 = px.scatter(
    km_df,
    x="log10km_mean",
    y="log10_value",
    title="log10km_mean vs log10_value"
    hover_data = ["substrate_smiles"]
)
fig2.show()
```
