```python
#kcat histogram 

import os
import numpy as np
import matplotlib.pyplot as plt

clip = 300

# Set working directory
path = "/Users/chideraangwadom/SerineProtease/molecule_analysis"
os.chdir(path)

file = "filtered_kcat_serine_proteases.csv"

# Load the data safely
try:
    print(f"📥 Reading file: {file}")
    data = np.genfromtxt(file, delimiter=',', skip_header=1, filling_values=np.nan)

    # Check if data has the right shape
    if data.shape[0] <= clip or data.shape[1] <= 1:
        print("❌ Not enough data after clipping.")
    else:
        wn = data[clip:, 0]  # x-axis: wavenumbers, etc.
        spectrum = data[clip:, 1]  # y-axis: intensity

        # Flatten and clean any NaNs
        spectrum = spectrum[~np.isnan(spectrum)]

        # Plot histogram
        plt.figure(figsize=(8, 6))
        plt.hist(spectrum, bins=30, color='mediumslateblue', alpha=0.8)
        plt.title("Histogram of Spectral Intensities\n(filtered_kcat_serine_proteases.csv)")
        plt.xlabel("Intensity")
        plt.ylabel("Frequency")
        plt.grid(True)
        plt.tight_layout()
        plt.show()

except Exception as e:
    print(f"❌ Error loading file: {e}")
```

```python
import pandas as pd
import matplotlib.piyplot as plt
import seaborn as sns
from scipy.stats import pearsonr

# Load re-uploaded files
kcat_df = pd.read_csv("/Users/chideraangwadom/SerineProtease/molecule_analysis/filtered_kcat_serine_proteases.csv")
km_df = pd.read_csv("/Users/chideraangwadom/SerineProtease/molecule_analysis/filtered_km_serine_proteases.csv")

# Extract the required columns
data = kcat_df[['log10kcat_max', 'reaction_mw_diff_perc']].dropna()

# Create scatter plot with hexbin, regression line, and correlation matrix
fig, ax = plt.subplots(2, 2, figsize=(14, 12), gridspec_kw={'width_ratios': [3, 1], 'height_ratios': [3, 1]})
((ax_scatter, ax_corr), (ax_hexbin, _)) = ax

# Scatter plot with regression line
sns.regplot(
    data=data,
    x='reaction_mw_diff_perc',
    y='log10kcat_max',
    ax=ax_scatter,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)
ax_scatter.set_title('Scatter Plot with Linear Regression')
ax_scatter.set_xlabel('reaction_mw_diff_perc')
ax_scatter.set_ylabel('log10kcat_max')

# Hexbin plot
hb = ax_hexbin.hexbin(
    data['reaction_mw_diff_perc'],
    data['log10kcat_max'],
    gridsize=30,
    cmap='viridis',
    mincnt=1
)
ax_hexbin.set_title('Hexbin Plot')
ax_hexbin.set_xlabel('reaction_mw_diff_perc')
ax_hexbin.set_ylabel('log10kcat_max')
fig.colorbar(hb, ax=ax_hexbin, label='Count')

# Correlation matrix
corr = data.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', ax=ax_corr, cbar=False)
ax_corr.set_title('Correlation Matrix')
ax_corr.set_xticklabels(ax_corr.get_xticklabels(), rotation=45)
ax_corr.set_yticklabels(ax_corr.get_yticklabels(), rotation=0)

plt.tight_layout()
plt.show()
```

```python
#log10kcat_max vs temperature scatter plot (add hexbin and correlation matrix)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr

# Load your dataset
df = pd.read_csv("filtered_kcat_serine_proteases.csv")

# Keep only the relevant columns and drop missing values
df_temp = df[['log10kcat_max', 'temperature']].dropna()

# --- 1. Correlation ---
r, p = pearsonr(df_temp['log10kcat_max'], df_temp['temperature'])
print(f"Pearson correlation: r = {r:.2f}, p = {p:.2e}")

# --- 2. Hexbin Plot ---
plt.figure(figsize=(8, 6))
plt.hexbin(df_temp['temperature'], df_temp['log10kcat_max'], gridsize=30, cmap='plasma', mincnt=1)
plt.colorbar(label='Point Density')
plt.xlabel('Temperature (°C)')
plt.ylabel('log10(kcat_max)')
plt.title('Hexbin: log10(kcat_max) vs Temperature')
plt.grid(True)
plt.tight_layout()
plt.show()

# --- 3. Scatter Plot with Regression Fit ---
plt.figure(figsize=(8, 6))
sns.regplot(
    x='temperature',
    y='log10kcat_max',
    data=df_temp,
    scatter_kws={'alpha': 0.4, 'color': 'navy'},
    line_kws={'color': 'darkorange'}
)
plt.xlabel('Temperature (°C)')
plt.ylabel('log10(kcat_max)')
plt.title(f'Scatter Plot with Fit: log10(kcat_max) vs Temperature\nr = {r:.2f}, p = {p:.2e}')
plt.grid(True)
plt.tight_layout()
plt.show()

# --- 4. Correlation Matrix Heatmap ---
plt.figure(figsize=(4, 3))
sns.heatmap(df_temp.corr(), annot=True, cmap='coolwarm', fmt='.2f', cbar=False)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
#log10kcat_max vs pH scatter plot, hexbin (because dense, and correlation matrix)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr

# Load the dataset
df = pd.read_csv("filtered_kcat_serine_proteases.csv")

# Drop rows missing key values
df = df[['log10kcat_max', 'ph']].dropna()

# Calculate Pearson correlation
r, p = pearsonr(df['log10kcat_max'], df['ph'])

# --- 1. Hexbin plot (if dense) ---
plt.figure(figsize=(8, 6))
plt.hexbin(df['ph'], df['log10kcat_max'], gridsize=30, cmap='plasma', mincnt=1)
plt.colorbar(label='Count')
plt.xlabel('pH')
plt.ylabel('log10(kcat_max)')
plt.title("Hexbin: log10(kcat_max) vs pH")
plt.grid(True)
plt.tight_layout()
plt.show()

# --- 2. Scatter with regression line ---
plt.figure(figsize=(8, 6))
sns.regplot(
    x='ph',
    y='log10kcat_max',
    data=df,
    scatter_kws={'alpha': 0.5, 'color': 'darkgreen'},
    line_kws={'color': 'red'}
)
plt.xlabel('pH')
plt.ylabel('log10(kcat_max)')
plt.title(f'Scatter + Fit: log10(kcat_max) vs pH\nPearson r = {r:.2f}, p = {p:.2e}')
plt.grid(True)
plt.tight_layout()
plt.show()

# --- 3. Correlation matrix heatmap ---
plt.figure(figsize=(4, 3))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f", cbar=False)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```
